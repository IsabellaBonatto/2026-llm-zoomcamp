# 4. With summary_length = long, roughly how many times more output tokens does multilingual_agent use compared to the short summary? 

```python
multilingual_agent_long = [
  {
    "tokenUsage.inputTokenCount": 282
  },
  {
    "tokenUsage.totalTokenCount": 767
  },
  {
    "tokenUsage.outputTokenCount": 197
  }
]
multilingual_agent_long
```

```python
multilingual_agent_short_dict = {}

for i in multilingual_agent_short:
    for key, value in i.items():
        multilingual_agent_short_dict[key] = value
```

```python
multilingual_agent_long_dict = {}

for i in multilingual_agent_long:
    for key, value in i.items():
        multilingual_agent_long_dict[key] = value
```

```python
short_output_token_count = multilingual_agent_short_dict['tokenUsage.outputTokenCount']
long_output_token_count = multilingual_agent_long_dict['tokenUsage.outputTokenCount']
```

```python
print("Comparison long-short:", long_output_token_count/short_output_token_count)
```

Answer: 2-5x more
